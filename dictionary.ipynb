```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sbn
```

```python
df = pd.read_csv("311_Dictionary.csv" , header=None,on_bad_lines='skip')
```

```python
df.columns
```

```python
df = df.iloc[:, :2]
df.columns = ["Field", "Value"]
```

```python
df = df.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
df.tail()
```

```python
df=df.apply(
    lambda col: col.str.replace(r"\s+","",regex=True).str.strip()
    if col.dtype=='object' else col
)
```

```python
# Group by 'field' and aggregate 'value'
grouped_df1 = df.groupby("Field")["Value"].sum().reset_index()

# Get top 15 by value
top_locations = grouped_df1.sort_values(by='value', ascending=False).head(15)

# Plot pie chart
fig, ax = plt.subplots()
ax.pie(top_locations["value"], labels=top_locations["field"], autopct='%1.1f%%')

plt.tight_layout()
plt.grid()
plt.show()
```
