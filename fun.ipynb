```python
def greet(name):
    return f"Hello, {name}!"
if __name__=="__main__":
    print(greet("World!"))
```

```python
def abcd(name):
    return f"Mera ALLAH mujy ZRR {name}"
if __name__=="__main__":
    print(abcd("kamyab kry ga "))
```

```python
def power(base,exponent):
    return base ** exponent
if __name__=="__main__":
    print(power(2, 3))
    
```

```python
def sum_all(*args):
    return sum(args)
if __name__=="__main__":
    print(sum_all(1,2,3,4))
```

```python
x = 10
def func():
    x = 5
    print(x)
func()
print(x)
```

```python
x = 10

def update():
    global x
    x = x + 5

update()
print(x)
```

```python
def outer():
    x = 10
    
    def inner():
        nonlocal x
        x += 5
        print(x)
    
    inner()
    print(x)

outer()
```
